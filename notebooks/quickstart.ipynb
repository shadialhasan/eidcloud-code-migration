Welcome to **EidCloud Code Migration** — the universal AST-driven code migration and syntax transformation engine in pure PHP.

This notebook demonstrates:
1. Environment setup and repository cloning.
2. Generating an interactive **migration plan** with AST token preservation.
3. Applying modern PHP 8.4 syntax refactoring with visual unified diffs.

```python
# Cell 1: Environment verification & setup
!php -v
!git clone https://github.com/eidcloud/eidcloud-code-migration.git /content/eidcloud-code-migration
%cd /content/eidcloud-code-migration
!php tests/run_tests.php
```

```python
# Cell 2: Create legacy PHP sample & Plan AST migration
sample_legacy_code = '''<?php
class AccountManager {
    /** @var string */
    private $username;

    public function getUserName($user) {
        return $user !== null ? $user->getName() : null;
    }

    public function getRoleLevel($role) {
        switch ($role) {
            case 'admin':
                return 1;
            case 'editor':
                return 2;
            default:
                return 0;
        }
    }

    public function formatList($items, $factor) {
        return array_map(function($x) use ($factor) {
            return $x * $factor;
        }, $items);
    }
}
'''

with open('/tmp/legacy_project.php', 'w') as f:
    f.write(sample_legacy_code)

print("=== PLANNING AST MIGRATION (DIFF ONLY) ===")
!php bin/eidcloud-migrate plan /tmp/legacy_project.php --rule=all
```

```python
# Cell 3: Apply AST migration with automatic backup and verify updated code
print("=== APPLYING AST CODE MIGRATION ===")
!php bin/eidcloud-migrate apply /tmp/legacy_project.php --rule=all --backup

print("\n=== RESULTING MODERN PHP 8.4 CODE ===")
with open('/tmp/legacy_project.php', 'r') as f:
    print(f.read())
```
